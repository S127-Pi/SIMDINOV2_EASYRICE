```python
import os 
```

```python
# Define the parent directory
directory = "/media/new_volumn/MP_India_Cleaned_Hovenet"

# Loop through all folder names in the directory
for folder_name in os.listdir(directory):
    old_path = os.path.join(directory, folder_name)
    
    # Ensure it's a folder
    if os.path.isdir(old_path) and len(folder_name) >= 7:
        new_name = "0000" + folder_name[-2:]  # Replace first 5 characters with '00000'
        new_path = os.path.join(directory, new_name)
        # Rename the folder
        os.rename(old_path, new_path)
        print(f"Renamed '{folder_name}' to '{new_name}'")
```
